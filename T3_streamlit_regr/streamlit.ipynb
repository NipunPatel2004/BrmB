```python
import streamlit as st
import pickle
import numpy as np

# Load model and scaler
model = pickle.load(open("random_forest_regressor.pkl", "rb"))
scaler = pickle.load(open("scaler.pkl", "rb"))

st.set_page_config(
    page_title="Car Price Predictor",
    page_icon="🚗",
    layout="centered"
)

st.title("🚗 Car Price Prediction")
st.write("Enter car specifications to estimate price.")

# User Inputs
enginesize = st.number_input("Engine Size", min_value=50, max_value=400, value=120)
horsepower = st.number_input("Horsepower", min_value=40, max_value=300, value=100)
curbweight = st.number_input("Curb Weight", min_value=1000, max_value=5000, value=2500)
carwidth = st.number_input("Car Width", min_value=50.0, max_value=80.0, value=65.0)
wheelbase = st.number_input("Wheel Base", min_value=80.0, max_value=130.0, value=95.0)
highwaympg = st.number_input("Highway MPG", min_value=10, max_value=60, value=30)
peakrpm = st.number_input("Peak RPM", min_value=4000, max_value=7000, value=5500)

if st.button("Predict Price"):

    features = np.array([[
        enginesize,
        horsepower,
        curbweight,
        carwidth,
        wheelbase,
        highwaympg,
        peakrpm
    ]])

    features = scaler.transform(features)

    prediction = model.predict(features)

    st.success(f"Estimated Car Price: ${prediction[0]:,.2f}")
```
